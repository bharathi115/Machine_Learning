```python
from sklearn.cluster import KMeans
df=pd.read_csv("autos.csv")
df
```

```python
print(df.columns)
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X=df[['symboling', 'num_of_doors',
       'wheel_base', 'length',
       'width', 'height', 'curb_weight', 'num_of_cylinders',
       'engine_size', 'bore', 'stroke', 'compression_ratio',
       'horsepower', 'peak_rpm', 'city_mpg', 'highway_mpg', 'price']]
X
```

```python
Features=['symboling', 'num_of_doors',
       'wheel_base', 'length',
       'width', 'height', 'curb_weight', 'num_of_cylinders',
       'engine_size', 'bore', 'stroke', 'compression_ratio',
       'horsepower', 'peak_rpm', 'city_mpg', 'highway_mpg', 'price']
```

```python
X = df[Features].apply(pd.to_numeric, errors="coerce")
```

```python
X = X.fillna(X.median())
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
model=KMeans(n_clusters=5,random_state=42)
model.fit(X)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X_scaled)
centroids = kmeans.cluster_centers_
```

```python
import matplotlib.pyplot as plt

plt.scatter(
    centroids[:, 0],centroids[:, 1],marker="X",s=200,color="red",edgecolor="black",
    label="Centroids")
plt.xlabel('Engine Size')
plt.ylabel('Horsepower')
plt.title('K-Means Clustering Data')
plt.show()
```

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=labels,
    cmap="viridis",
    s=50,
    alpha=0.7
)
```

```python
data = pd.read_csv("Mall_Customers.csv")
data
```

```python
X=data[['CustomerID', 'Genre', 'Age', 'Annual Income (k$)',
       'Spending Score (1-100)']]
```

```python
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#Heirachical Clustering
hc = AgglomerativeClustering(n_clusters=5, linkage='ward')
data['Cluster'] = hc.fit_predict(X_scaled)
```

```python
#Scatter Chart
plt.figure(figsize=(9, 6))
plt.scatter(
    data['Annual Income (k$)'],
    data['Spending Score (1-100)'],
    c=data['Cluster'],
    cmap='viridis',
    s=60
)
```

```python
plt.figure(figsize=(12, 6))
linked = linkage(X_scaled, method='ward')
dendrogram(linked)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Customers')
plt.ylabel('Distance')
plt.show()
```
