```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

```python
data=pd.read_csv("disease_results.csv")
data
```

```python
le=LabelEncoder()
x=le.fit_transform(data["Actual"])
y=le.fit_transform(data["Predicted"])
```

```python
x
```

```python
y
```

```python
accuracy=accuracy_score(x,y)
precision=precision_score(x,y)
recall=recall_score(x,y)
f1=f1_score(x,y)
cm=confusion_matrix(x,y)
```

```python
print("accuracy")
print(accuracy)
print("precision")
print(precision)
print("recall")
print(recall)
print("f1")
print(f1)
print("confusion matrix")
print(cm)
```

```python
import pandas as pd
data=pd.read_csv("food_spicy_cheese.csv")
data
```

```python
from sklearn.naive_bayes import BernoulliNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
spicy_food=data["Spicy"]
cheese_food=data["Cheese"]
x=np.column_stack([spicy_food,cheese_food])
x
```

```python
y=le.fit_transform(data["Likes Food"])
y
```

```python
x_train,y_train,x_test,y_test= train_test_split(x,y,test_size=0.2)
```

```python
model=BernoulliNB()
model.fit(x_train,x_test)
```

```python
spicy_food=int(input("enter your answer(1-yes,no-0):"))
cheese_food=int(input("enter your answer(1-yes,no-0):"))
test_food=np.array([spicy_food,cheese_food]).reshape(1,-1)
test_food
answer=model.predict(test_food)
print(answer)
print(le.inverse_transform(answer))

```

```python
y_pred=model.predict(y_train)
y_pred
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
f1=f1_score(y_test,y_pred)
f1
```

```python
recall=recall_score(y_test,y_pred)
recall
```

```python
accuracy=accuracy_score(y_test,y_pred)
accuracy
```
